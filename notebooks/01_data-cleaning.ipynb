# Limpeza e Tratamento de Dados

Notebook responsável pela limpeza, padronização e transformação da base bruta de comportamento de compra de clientes.

**Entrada:** `data/raw/customer_shopping_behavior.csv`  
**Saída:** `data/processed/customer_shopping_behavior_clean.csv`

```python
import pandas as pd
```

```python
df = pd.read_csv(r'../data/raw/customer_shopping_behavior.csv', sep=',')
df.head()
```

### Tratamento de Valores Ausentes — `review_rating`

**Estratégia:** imputação por grupo, utilizando a mediana de `review_rating` calculada individualmente para cada `category`.

**Justificativa:** decidi pela imputação por agrupamento, pois não assume homogeneidade entre subgrupos da base. Ao contrário da imputação global, que aplica uma única estatística a todos os registros. Essa abordagem preserva eventuais diferenças de distribuição entre categorias, sem custo adicional relevante de implementação.

```python
# Imputando os valores ausentes de avaliações pela mediana de cada categoria
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.mean()))
```

```python
# Verificando novamente se ainda tem valores ausentes na coluna 'Review Rating'
df['Review Rating'].isna().sum()
```

```python
# Padronizando colunas no formato snake_case:
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ', '_')
df = df.rename(columns={'purchase_amount_(usd)': 'purchase_amount'})
```

```python
df.columns
```

### Segmentação de Clientes por Faixa Etária — `age_group`

**Estratégia:** divisão da variável `age` em 4 grupos utilizando `pd.qcut` (segmentação por quantil).

**Justificativa:** `qcut` particiona os dados em grupos de tamanho aproximadamente igual, com base na distribuição real da variável — ao contrário de `cut`, que utiliza intervalos fixos e arbitrários. Essa escolha garante que cada segmento seja estatisticamente representativo da base, refletindo a distribuição observada em vez de faixas etárias pré-definidas.

```python
# Agrupando os clietes em 4 faixas etárias
labels = ['Young Adult', 'Adult', 'Middle-Agged', 'Senior']

# Convertendo as idades em dados categóricos e criando 4 grupos de tamanho iguais com base na distrbuição de dados
df['age_group'] = pd.qcut(df['age'], q=4, labels=labels)
```

```python
df[['age_group', 'age']].head(10)
```

### Conversão de Frequência de Compra — `purchase_frequency_days`

**Estratégia:** mapeamento dos rótulos textuais de `frequency_of_purchases` (ex: "Fortnightly", "Monthly") para valores numéricos em dias.

**Justificativa:** a coluna original é categórica e não permite operações quantitativas diretas. A conversão para dias viabiliza análises como correlação e agregações estatísticas sobre o comportamento de recorrência de compra. Os valores atribuídos seguem o significado padrão de cada termo em inglês.

```python
# Dicionária que associa 'frequency_of_purchases' (str) com números de dias correspondentes
frequency_mapping = {
    'Fortnightly': 14,
    'Weekly': 7,
    'Monthly': 30,
    'Quarterly': 90,
    'Bi-Weekly': 14,
    'Annually': 365,
    'Every 3 Months': 90
}

# Criando uma coluna para frequência de compras por dia trasnformando em valores numéricos
df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```

```python
df[['purchase_frequency_days', 'frequency_of_purchases']].head(10)
```

### Verificação de Redundância — `discount_applied` x `promo_code_used`

**Estratégia:** comparação direta entre as colunas `discount_applied` e `promo_code_used` para identificar se carregam a mesma informação.

**Justificativa:** manter colunas redundantes aumenta a dimensionalidade da base sem agregar informação nova, além de poder introduzir ambiguidade em análises futuras. A comparação confirmou equivalência total entre as colunas.

```python
df[['discount_applied', 'promo_code_used']].head(10)
```

```python
# Verificando redundancia das colunas 'discount_applied' e 'promo_code_used'
(df['discount_applied'] == df['promo_code_used']).all()
```

```python
# Removendo a coluna de código promocional usado
df = df.drop('promo_code_used', axis=1)
```

```python
df.columns
```

### Exportação do Dataset Tratado

**Estratégia:** exportação do DataFrame limpo para `data/processed/customer_shopping_behavior_clean.csv`.

**Justificativa:** manter dado bruto (`data/raw/`) e dado tratado (`data/processed/`) em diretórios separados preserva a rastreabilidade do pipeline e permite reprocessar os dados a partir da fonte original, se necessário.

```python
# Salvando o DataFrame após a limpeza dos dados
df.to_csv(r"../data/processed/customer_shopping_behavior_clean.csv", sep=",", index=False)
```
