# Conexão e Carga no PostgreSQL

Notebook responsável por estabelecer conexão com o banco de dados PostgreSQL e carregar a base tratada na tabela `customer`, utilizada como fonte para as análises SQL em `queries/`.

**Entrada:** `data/processed/customer_shopping_behavior_clean.csv`  
**Saída:** tabela `customer` no banco de dados PostgreSQL  

```python
import os

import pandas as pd
from dotenv import load_dotenv
from sqlalchemy import create_engine
```

### Carregamento de Credenciais

**Estratégia:** credenciais de acesso ao banco armazenadas em variáveis de ambiente (`.env`), carregadas via `python-dotenv`.

**Justificativa:** evita expor credenciais sensíveis diretamente no código-fonte, prática essencial para projetos versionados publicamente (o arquivo `.env` é excluído do repositório via `.gitignore`).

```python
# Carregando as variáveis de ambiente
load_dotenv()

# Definindo as credenciais e parâmetros de conexão do PostgreSQL
username = os.getenv("DB_USERNAME")
password = os.getenv("DB_PASSWORD")
host = os.getenv("DB_HOST")
port = os.getenv("DB_PORT")
database = os.getenv("DB_DATABASE")
```

```python
# Importando o DF limpo
df = pd.read_csv('../data/processed/customer_shopping_behavior_clean.csv', sep=',')
```

### Conexão e Carga dos Dados

**Estratégia:** conexão via SQLAlchemy (`create_engine`) e carga da tabela `customer` utilizando `if_exists="replace"`.

**Justificativa:** a estratégia `replace` garante que a tabela sempre reflita a versão mais recente do dataset tratado, evitando duplicação de registros em re-execuções do notebook,.

```python
# Criando a conexão ao banco de Dados
engine = create_engine(f"postgresql+psycopg2://{username}:{password}@{host}:{port}/{database}")

# Carregando o para o Banco de Dados no PostgreSQL
try:
    table_name = "customer"
    df.to_sql(table_name, con=engine, if_exists="replace", index=False)

    print(f"Os dados foram carregados com sucesso na tabela '{table_name}' no banco de dados '{database}'.")

except Exception as e:
    print(f"Erro ao carregar os dados: {e}")
```
